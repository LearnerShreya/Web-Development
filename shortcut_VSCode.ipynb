### **📝 Essential VS Code Shortcuts (Windows)**  

| **Category**          | **Shortcut**                            | **Function** |
|----------------------|--------------------------------------|-------------|
| **🔹 General**       | `Ctrl + Shift + P / F1`             | Open Command Palette |
|                      | `Ctrl + P`                          | Quick open file |
|                      | `Ctrl + ,`                          | Open settings |
|                      | `Ctrl + B`                          | Toggle sidebar |
|                      | `` Ctrl + ` ``                      | Toggle terminal |
|                      | `Ctrl + N`                          | New file |
|                      | `Ctrl + S`                          | Save file |
|                      | `Ctrl + Shift + S`                  | Save As |
|                      | `Ctrl + W`                          | Close file |
|                      | `Ctrl + Shift + T`                  | Reopen last closed file |
|                      | `Ctrl + K, Ctrl + S`                | Open keyboard shortcuts |
| **✏️ Editing**       | `Ctrl + X`                          | Cut line |
|                      | `Ctrl + C`                          | Copy line |
|                      | `Ctrl + V`                          | Paste |
|                      | `Ctrl + Shift + K`                  | Delete line |
|                      | `Ctrl + Enter`                      | Insert new line below |
|                      | `Ctrl + Shift + Enter`              | Insert new line above |
|                      | `Alt + ↑ / ↓`                       | Move line up/down |
|                      | `Shift + Alt + ↑ / ↓`               | Duplicate line up/down |
|                      | `Ctrl + D`                          | Select next occurrence of word |
|                      | `Ctrl + Shift + L`                  | Select all occurrences |
|                      | `Ctrl + L`                          | Select entire line |
|                      | `Ctrl + /`                          | Toggle comment |
|                      | `Shift + Alt + A`                   | Toggle block comment |
|                      | `Ctrl + Backspace`                  | Delete previous word |
|                      | `Ctrl + Delete`                     | Delete next word |
| **🔍 Navigation**     | `Ctrl + Shift + O`                  | Go to symbol in file |
|                      | `Ctrl + T`                          | Go to symbol in workspace |
|                      | `Ctrl + G`                          | Jump to line |
|                      | `Ctrl + Shift + Tab`                | Cycle through open files |
|                      | `Ctrl + Tab`                        | Switch between open files |
|                      | `Alt + ← / →`                       | Navigate back/forward |
|                      | `Ctrl + F`                          | Find |
|                      | `Ctrl + H`                          | Find and replace |
|                      | `F3 / Shift + F3`                   | Find next/previous match |
|                      | `Ctrl + Shift + F`                  | Find in all files |
|                      | `Ctrl + Shift + H`                  | Replace in all files |
| **🖱️ Multi-cursor**   | `Alt + Click`                      | Add additional cursor |
|                      | `Ctrl + Alt + ↑ / ↓`               | Add cursor above/below |
|                      | `Shift + Alt + →`                  | Expand selection right |
|                      | `Shift + Alt + ←`                  | Shrink selection left |
|                      | `Ctrl + Shift + L`                  | Select all occurrences |
|                      | `Ctrl + Shift + Alt + ↑ / ↓`       | Add column selection |
| **🛠️ Formatting**    | `Alt + Shift + F`                  | Format document |
|                      | `Ctrl + K, Ctrl + F`                | Format selection |
|                      | `Ctrl + Shift + M`                  | Show problems (errors, warnings) |
|                      | `F2`                                | Rename symbol |
|                      | `Ctrl + Shift + R`                  | Refactor |
|                      | `Ctrl + K, Ctrl + C`                | Add line comment |
|                      | `Ctrl + K, Ctrl + U`                | Remove line comment |
| **📂 File & Workspace** | `Ctrl + K, Ctrl + O`             | Open folder |
|                      | `Ctrl + K, Ctrl + W`                | Close all editors |
|                      | `Ctrl + Shift + N`                  | Open new VS Code window |
|                      | `Ctrl + K, Ctrl + X`                | Close all files except current |
|                      | `Ctrl + Shift + E`                  | Show Explorer |
|                      | `Ctrl + Shift + D`                  | Open Run and Debug |
|                      | `Ctrl + J`                          | Toggle panel (terminal, problems, output, etc.) |
| **🔧 Debugging**      | `F5`                                | Start debugging |
|                      | `Shift + F5`                        | Stop debugging |
|                      | `F9`                                | Toggle breakpoint |
|                      | `F10`                               | Step over |
|                      | `F11`                               | Step into |
|                      | `Shift + F11`                       | Step out |
| **💡 Extensions & Git** | `Ctrl + Shift + X`               | Open Extensions Marketplace |
|                      | `Ctrl + Shift + G`                  | Open Source Control (Git) |
|                      | `Ctrl + K, Ctrl + U`                | Undo last Git commit |
|                      | `Ctrl + K, Ctrl + P`                | Push Git changes |
|                      | `Ctrl + K, Ctrl + O`                | Pull Git changes |
