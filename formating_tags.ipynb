# HTML Text Formatting Tags

## 1. **Bold Text**
- **Tags**: `<b>` and `<strong>`
- **Purpose**: Both tags make text bold, but their usage differs:
  - **`<b>`**: Used purely for styling purposes and does not indicate importance.
  - **`<strong>`**: Semantically highlights text as important and meaningful.
- **Example**:
  ```html
  <!-- Bold for style -->
  <b>This text is bold for styling purposes.</b>
  
  <!-- Bold for importance -->
  <strong>This text is bold because it is important.</strong>
  ```

  **Output**:
  **This text is bold for styling purposes.**  
  **This text is bold because it is important.**

---

## 2. **Italic Text**
- **Tags**: `<i>` and `<em>`
- **Purpose**:
  - **`<i>`**: Used for stylistic italicization, such as foreign words or technical terms.
  - **`<em>`**: Indicates emphasis, adding semantic meaning to the text.
- **Example**:
  ```html
  <!-- Italic for style -->
  <i>This text is italicized for styling.</i>

  <!-- Italic for emphasis -->
  <em>This text is emphasized and italicized.</em>
  ```

  **Output**:  
  *This text is italicized for styling.*  
  *This text is emphasized and italicized.*

---

## 3. **Underlined Text**
- **Tag**: `<u>`
- **Purpose**: Underlines text to emphasize or highlight it.
- **Example**:
  ```html
  <u>This text is underlined for emphasis.</u>
  ```

  **Output**:  
  <u>This text is underlined for emphasis.</u>

---

## 4. **Strikethrough Text**
- **Tags**: `<s>` and `<del>`
- **Purpose**:
  - **`<s>`**: Shows text that is no longer accurate or relevant but not removed.
  - **`<del>`**: Denotes deleted content, often used in document revisions.
- **Example**:
  ```html
  <s>This text is outdated or irrelevant.</s>
  <del>This text was removed from the document.</del>
  ```

  **Output**:  
  ~~This text is outdated or irrelevant.~~  
  ~~This text was removed from the document.~~

---

## 5. **Subscript and Superscript**
- **Tags**: `<sub>` and `<sup>`
- **Example**:
  ```html
  H<sub>2</sub>O (Water)
  x<sup>2</sup> (Squared)
  ```

  **Output**:  
  H₂O (Water)  
  x² (Squared)

---

## 6. **Highlighted Text**
- **Tag**: `<mark>`
- **Example**:
  ```html
  <mark>This text is highlighted for emphasis.</mark>
  ```

  **Output**:  
  <mark>This text is highlighted for emphasis.</mark>

---

## 7. **Small Text**
- **Tag**: `<small>`
- **Example**:
  ```html
  <small>This text is smaller in size.</small>
  ```

  **Output**:  
  <small>This text is smaller in size.</small>

---

## 8. **Monospaced or Code Text**
- **Tags**: `<code>` and `<pre>`
- **Example**:
  ```html
  <code>print("Hello, World!")</code>
  <pre>
  def hello_world():
      print("Hello, World!")
  </pre>
  ```

  **Output**:  
  `print("Hello, World!")`  
  ```  
  def hello_world():
      print("Hello, World!")
  ```

---

## 9. **Quotation Text**
- **Tags**: `<q>` and `<blockquote>`
- **Example**:
  ```html
  <q>The journey of a thousand miles begins with a single step.</q>
  <blockquote>
      This is a longer block quote. It can span multiple lines and is often used to cite sources.
  </blockquote>
  ```

  **Output**:  
  “The journey of a thousand miles begins with a single step.”  
  > This is a longer block quote. It can span multiple lines and is often used to cite sources.

---

## 10. **Inserted and Deleted Text**
- **Tags**: `<ins>` and `<del>`
- **Example**:
  ```html
  <ins>This text was added to the document.</ins>
  <del>This text was removed from the document.</del>
  ```

  **Output**:  
  <ins>This text was added to the document.</ins>  
  ~~This text was removed from the document.~~

---

## 11. **Abbreviations**
- **Tag**: `<abbr>`
- **Example**:
  ```html
  <abbr title="HyperText Markup Language">HTML</abbr>
  ```

  **Output**:  
  <abbr title="HyperText Markup Language">HTML</abbr> (Hover over it to see full form)

---

## 12. **Text with Direction**
- **Tag**: `<bdo>`
- **Example**:
  ```html
  <bdo dir="rtl">This text is reversed.</bdo>
  ```

  **Output**:  
  ןoitisop si txet sihT

---

## 13. **Citation or Reference**
- **Tag**: `<cite>`
- **Example**:
  ```html
  <cite>The Great Gatsby</cite> by F. Scott Fitzgerald
  ```

  **Output**:  
  *The Great Gatsby* by F. Scott Fitzgerald

---

## **Summary**
HTML formatting tags enhance readability and emphasize content. Use semantic tags like `<strong>`, `<em>`, `<mark>`, and `<cite>` for meaning, while `<b>`, `<i>`, and `<u>` help with visual styling. Proper usage ensures accessibility and clarity.
