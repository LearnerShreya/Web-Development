## **📌 VS Code HTML Shortcuts (Emmet Abbreviations) 🖥️**  

### **1️⃣ Basic Boilerplate**
| **Shortcut** | **Description** | **Output** |
|-------------|---------------|------------|
| `! + Enter` | Generates a complete HTML5 boilerplate | Full HTML structure |

---

### **2️⃣ Creating Elements**
| **Shortcut** | **Description** | **Example** | **Output** |
|-------------|---------------|------------|------------|
| `element` | Creates a single element | `div` | `<div></div>` |
| `element*number` | Creates multiple elements | `p*3` | `<p></p><p></p><p></p>` |
| `element{name}` | Inserts text inside a tag | `h1{Hello World}` | `<h1>Hello World</h1>` |

---

### **3️⃣ Nesting and Grouping Elements**
| **Shortcut** | **Description** | **Example** | **Output** |
|-------------|---------------|------------|------------|
| `parent>child` | Creates nested elements | `div>ul>li` | `<div><ul><li></li></ul></div>` |
| `sibling + sibling` | Creates sibling elements | `h1+p` | `<h1></h1><p></p>` |
| `grouping ( )` | Groups elements logically | `(div>p)+ul>li*3` | `<div><p></p></div><ul><li></li><li></li><li></li></ul>` |

---

### **4️⃣ Adding Classes & IDs**
| **Shortcut** | **Description** | **Example** | **Output** |
|-------------|---------------|------------|------------|
| `element.class` | Assigns a class | `div.container` | `<div class="container"></div>` |
| `element#id` | Assigns an ID | `div#main` | `<div id="main"></div>` |
| `element.class1.class2` | Multiple classes | `button.btn.primary` | `<button class="btn primary"></button>` |

---

### **5️⃣ Adding Attributes**
| **Shortcut** | **Description** | **Example** | **Output** |
|-------------|---------------|------------|------------|
| `element[attr=value]` | Adds an attribute | `input[type=text]` | `<input type="text">` |
| `element[attr1=value1 attr2=value2]` | Multiple attributes | `img[src="image.jpg" alt="Photo"]` | `<img src="image.jpg" alt="Photo">` |
| `a[href="https://example.com"]{Visit}` | Creates a link with text | `a[href="https://google.com"]{Google}` | `<a href="https://google.com">Google</a>` |

---

### **6️⃣ Lists & Tables**
| **Shortcut** | **Description** | **Example** | **Output** |
|-------------|---------------|------------|------------|
| `ul>li*3` | Creates an unordered list | `ul>li*3` | `<ul><li></li><li></li><li></li></ul>` |
| `table>tr*2>td*3` | Creates a table | `table>tr*2>td*3` | `<table><tr><td></td><td></td><td></td></tr><tr><td></td><td></td><td></td></tr></table>` |

---

### **7️⃣ Forms & Inputs**
| **Shortcut** | **Description** | **Example** | **Output** |
|-------------|---------------|------------|------------|
| `form>input+button{Submit}` | Creates a basic form | `form>input[type=text]+button{Submit}` | `<form><input type="text"><button>Submit</button></form>` |
| `input:email` | Creates an email input field | `input:email` | `<input type="email">` |
| `input:password` | Creates a password input field | `input:password` | `<input type="password">` |
| `textarea` | Creates a textarea | `textarea` | `<textarea></textarea>` |

---

### **8️⃣ Media Elements**
| **Shortcut** | **Description** | **Example** | **Output** |
|-------------|---------------|------------|------------|
| `img[src="image.jpg" alt="Description"]` | Creates an image | `img[src="logo.png" alt="Company Logo"]` | `<img src="logo.png" alt="Company Logo">` |
| `video[src="video.mp4" controls]` | Creates a video element | `video[src="movie.mp4" controls]` | `<video src="movie.mp4" controls></video>` |
| `audio[src="sound.mp3" controls]` | Creates an audio element | `audio[src="song.mp3" controls]` | `<audio src="song.mp3" controls></audio>` |

---

### **9️⃣ Meta & Links**
| **Shortcut** | **Description** | **Example** | **Output** |
|-------------|---------------|------------|------------|
| `meta:viewport` | Adds viewport meta tag | `meta:viewport` | `<meta name="viewport" content="width=device-width, initial-scale=1.0">` |
| `link:css` | Creates a CSS link | `link:css` | `<link rel="stylesheet" href="style.css">` |
| `script:src` | Creates a script tag | `script:src` | `<script src="script.js"></script>` |

---

### **🔹 Bonus: Custom Layouts**
| **Shortcut** | **Description** | **Example** | **Output** |
|-------------|---------------|------------|------------|
| `div>(header>h1{Title})+main+footer` | Creates a structured layout | `div>(header>h1{Title})+main+footer` | `<div><header><h1>Title</h1></header><main></main><footer></footer></div>` |
| `section>article*2>h2{Heading}+p{Text}` | Creates a blog-like section | `section>article*2>h2{Title}+p{Content}` | `<section><article><h2>Title</h2><p>Content</p></article><article><h2>Title</h2><p>Content</p></article></section>` |
